## 1. Bloch KüresiTek qubit'in durumunu 3D bir küre üzerinde görselleştirebiliriz.$$|\psi\rangle = \cos\frac{\theta}{2}|0\rangle + e^{i\phi}\sin\frac{\theta}{2}|1\rangle$$- **θ** (theta): Kuzey kutbu (|0⟩) ile güney kutbu (|1⟩) arasındaki açı- **φ** (phi): Ekvator düzlemindeki faz açısı### Önemli Noktalar:- **Kuzey kutbu** (θ=0): ∣0⟩- **Güney kutbu** (θ=π): ∣1⟩- **Ekvator** (θ=π/2): Süperpozisyon durumları (∣+⟩, ∣−⟩, ∣i⟩, ∣−i⟩)- **Kürenin içi**: Karışık durumlar (mixed states) — ders kapsamı dışı

## 2. SüperpozisyonKuantum mekaniğinin en temel özelliği: bir qubit aynı anda **hem 0 hem 1** olabilir.$$|\psi\rangle = \alpha|0\rangle + \beta|1\rangle$$Bu bir "kararsızlık" değil, gerçek bir fiziksel durumdur. Ölçüm yapana kadar qubit gerçekten her iki durumda birden var.### AnalojiBir yazı-tura. Havadayken hem yazı hem tura. Yere düşünce (=ölçüm) birini görürsün. Ama havadayken gerçekten ikisi birden.

```python
import numpy as npimport matplotlib.pyplot as plt# Süperpozisyon — farklı α, β değerleri için olasılıklarfig, axes = plt.subplots(1, 4, figsize=(14, 4))states_info = [    ('|0⟩', 1, 0),    ('|+⟩ = (|0⟩+|1⟩)/√2', 1/np.sqrt(2), 1/np.sqrt(2)),    ('0.95|0⟩ + 0.31|1⟩', 0.95, 0.31),    ('|1⟩', 0, 1),]for ax, (name, a, b) in zip(axes, states_info):    probs = [abs(a)**2, abs(b)**2]    bars = ax.bar(['|0⟩', '|1⟩'], probs, color=['#3b82f6', '#ef4444'], alpha=0.85, edgecolor='white', width=0.6)    ax.set_ylim(0, 1.15)    ax.set_title(name, fontsize=11, fontweight='bold')    for bar, p in zip(bars, probs):        if p > 0.01:            ax.text(bar.get_x() + bar.get_width()/2, p + 0.03, f'%{p*100:.0f}',                     ha='center', fontsize=12, fontweight='bold')    ax.axhline(y=0.5, color='gray', linestyle='--', alpha=0.3)    ax.set_ylabel('Olasılık' if ax==axes[0] else '', fontsize=10)plt.suptitle('Süperpozisyon — Farklı Durumların Olasılık Dağılımları', fontsize=14, fontweight='bold', y=1.02)plt.tight_layout()plt.show()
```

## 3. ÖlçümÖlçüm, süperpozisyonu **çökertir**. Sonuç her zaman kesin: 0 veya 1.$$|\psi\rangle = \alpha|0\rangle + \beta|1\rangle \xrightarrow{\text{ölçüm}} \begin{cases} |0\rangle & \text{olasılık } |\alpha|^2 \\ |1\rangle & \text{olasılık } |\beta|^2 \end{cases}$$### Born Kuralı$P(\text{sonuç } = 0) = |\alpha|^2$, $P(\text{sonuç } = 1) = |\beta|^2$### Ölçümün Geri Dönüşü Yok!Ölçüm sonrası durum **kalıcı olarak** değişir. Tekrar ölçsen hep aynı sonucu alırsın.

## 4. Başka ihtimal yok."### Dolanıklık ≠ Klasik Korelasyon**Klasik korelasyon (çorap analojisi):** Bir çift çorabı iki kutuya koy. Bir kutuyu İstanbul'a, birini Ankara'ya gönder. İstanbul'daki kutuyu aç, kırmızı çorap gör → Ankara'da mavi olduğunu BİLİRSİN. Ama bu sürpriz değil, baştan belliydi.**Kuantum dolanıklık:** Çorap kutuya konulana kadar rengi YOK. Kutuyu açtığın an renk belirleniyor — ve diğer kutu da **anında** karşıt rengi alıyor, mesafe fark etmez. Einstein buna "ürkütücü uzaktan etki" (spooky action at a distance) dedi.
