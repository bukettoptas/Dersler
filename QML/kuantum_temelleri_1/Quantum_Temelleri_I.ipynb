## 1. Moore Yasası

Gordon Moore (1965): Entegre devredeki transistör sayısı her **2 yılda 2 katına** çıkar. Transistörler küçüldükçe atom boyutuna yaklaşıyoruz → klasik fizik bozuluyor → **kuantum bilgisayar**ın motivasyonu burada başlıyor.

```python
import numpy as np
import matplotlib.pyplot as plt

yillar = [1971, 1978, 1985, 1993, 2000, 2008, 2016, 2024]
transistor = [2.3e3, 2.9e4, 2.75e5, 3.1e6, 4.2e7, 7.3e8, 1.75e10, 1.14e11]
isimler = ['4004', '8086', '386', 'Pentium', 'P4', 'Core i7', 'Apple A10', 'M3 Ultra']

fig, ax = plt.subplots(figsize=(11, 5))
ax.semilogy(yillar, transistor, 'o-', color='#8b5cf6', lw=2.5, ms=10, zorder=3)

# Her noktaya isim yaz
for y, t, n in zip(yillar, transistor, isimler):
    ax.annotate(n, (y, t), textcoords="offset points", xytext=(0, 14),
                ha='center', fontsize=8, color='#64748b')

# Moore tahmini
t_theory = transistor[0] * 2**((np.array(yillar) - yillar[0]) / 2)
ax.semilogy(yillar, t_theory, '--', color='#ef4444', alpha=0.5, label='Moore tahmini')

ax.axhline(y=1e12, color='#f59e0b', ls=':', alpha=0.7)
ax.text(2018, 2e12, '⚠️ Atom boyutu sınırı', fontsize=11, color='#f59e0b', fontweight='bold')

ax.set_xlabel('Yıl', fontsize=12)
ax.set_ylabel('Transistör Sayısı (log)', fontsize=12)
ax.set_title("Moore Yasası — Transistör Sayısının Üstel Artışı", fontsize=14, fontweight='bold')
ax.legend(fontsize=10)
ax.grid(True, alpha=0.2)
plt.tight_layout()
plt.show()
```

## 3. Dirac Notasyonu

Paul Dirac'ın icadı — vektörleri pratik yazmak için. **Ket** = sütun vektör, **Bra** = satır vektör.

$$|0\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix}, \quad |1\rangle = \begin{pmatrix} 0 \\ 1 \end{pmatrix}, \quad |\psi\rangle = \alpha|0\rangle + \beta|1\rangle \quad (|\alpha|^2 + |\beta|^2 = 1)$$

```python
import numpy as np

# === KET ve BRA tanımla ===
ket_0 = np.array([[1], [0]])   # |0⟩
ket_1 = np.array([[0], [1]])   # |1⟩

bra_0 = ket_0.conj().T         # ⟨0|
bra_1 = ket_1.conj().T         # ⟨1|

print("=== Bazlar ===")
print(f"|0⟩ = {ket_0.flatten()}")
print(f"|1⟩ = {ket_1.flatten()}")
print(f"⟨0| = {bra_0.flatten()}")
print(f"⟨1| = {bra_1.flatten()}")
```

```python
# === Süperpozisyon durumu ===
alpha = 1 / np.sqrt(2)
beta  = 1 / np.sqrt(2)
psi = alpha * ket_0 + beta * ket_1

print("=== Süperpozisyon ===")
print(f"|ψ⟩ = {alpha:.4f}|0⟩ + {beta:.4f}|1⟩")
print(f"|ψ⟩ = {psi.flatten()}")
print(f"Normalizasyon: |α|² + |β|² = {abs(alpha)**2 + abs(beta)**2:.1f} ✅")
```

```python
# === Karmaşık sayılı örnek ===
alpha_c = (1 + 1j) / 2
beta_c  = (1 - 1j) / 2
psi_c = alpha_c * ket_0 + beta_c * ket_1

print("=== Karmaşık Genlikli Durum ===")
print(f"α = {alpha_c},  β = {beta_c}")
print(f"|ψ⟩ = {psi_c.flatten()}")
print(f"|α|² = {abs(alpha_c)**2:.2f},  |β|² = {abs(beta_c)**2:.2f}")
print(f"Toplam = {abs(alpha_c)**2 + abs(beta_c)**2:.1f} ✅")
```

## 4. İç Çarpım ve Dış Çarpım

**İç çarpım** ⟨φ|ψ⟩ → skaler (tek sayı) → **benzerlik ölçer** → Born kuralı: P = |⟨φ|ψ⟩|²

**Dış çarpım** |ψ⟩⟨φ| → matris → **operatör oluşturur** → projeksiyon

```python
# === İÇ ÇARPIM ===
print("=== İç Çarpım (Inner Product) ===")
print(f"⟨0|0⟩ = {float((bra_0 @ ket_0)[0,0]):.0f}   (aynı → 1)")
print(f"⟨1|1⟩ = {float((bra_1 @ ket_1)[0,0]):.0f}   (aynı → 1)")
print(f"⟨0|1⟩ = {float((bra_0 @ ket_1)[0,0]):.0f}   (dik  → 0)")
print(f"⟨1|0⟩ = {float((bra_1 @ ket_0)[0,0]):.0f}   (dik  → 0)")
```

```python
# === Born Kuralı ile Olasılık ===
psi_plus = (ket_0 + ket_1) / np.sqrt(2)   # |+⟩

p0 = abs(bra_0 @ psi_plus)[0,0] ** 2
p1 = abs(bra_1 @ psi_plus)[0,0] ** 2

print("=== Born Kuralı ===")
print(f"|+⟩ = (|0⟩+|1⟩)/√2")
print(f"P(0) = |⟨0|+⟩|² = {p0:.2f} → %{p0*100:.0f}")
print(f"P(1) = |⟨1|+⟩|² = {p1:.2f} → %{p1*100:.0f}")
```

```python
# === DIŞ ÇARPIM ===
proj_0 = ket_0 @ bra_0   # |0⟩⟨0|
proj_1 = ket_1 @ bra_1   # |1⟩⟨1|

print("=== Dış Çarpım (Outer Product) ===")
print(f"|0⟩⟨0| =\n{proj_0}\n")
print(f"|1⟩⟨1| =\n{proj_1}\n")
print(f"|0⟩⟨0| + |1⟩⟨1| = I =\n{proj_0 + proj_1}")
```

## 6. Hilbert Oteli ve Hilbert Uzayı

### Hilbert Oteli
Sonsuz odalı otel, tüm odalar dolu. Yeni misafir → herkes n→n+1'e kayar → Oda 1 boşalır. **∞+1=∞**

Sonsuz otobüs → herkes n→2n'e kayar → tek odalar boşalır. **∞+∞=∞**

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

fig, axes = plt.subplots(2, 1, figsize=(12, 5))

# Çözüm 1: n → n+1
ax = axes[0]
ax.set_xlim(-0.5, 11); ax.set_ylim(-0.3, 1.8)
ax.axis('off')
ax.set_title('Çözüm 1: n → n+1  (∞ + 1 = ∞)', fontsize=13, fontweight='bold', color='#0891b2')

for i in range(8):
    c = '#10b981' if i == 0 else '#3b82f6'
    lbl = 'BOŞ!' if i == 0 else f'{i}→{i+1}'
    rect = patches.FancyBboxPatch((i*1.3, 0.6), 1.1, 0.7, boxstyle="round,pad=0.1",
                                   facecolor=c, alpha=0.8, edgecolor='white', lw=1.5)
    ax.add_patch(rect)
    ax.text(i*1.3+0.55, 0.95, lbl, ha='center', va='center', fontsize=10, color='white', fontweight='bold')
    ax.text(i*1.3+0.55, 0.35, f'Oda {i+1}', ha='center', fontsize=8, color='#94a3b8')

# Çözüm 2: n → 2n
ax2 = axes[1]
ax2.set_xlim(-0.5, 11); ax2.set_ylim(-0.3, 1.8)
ax2.axis('off')
ax2.set_title('Çözüm 2: n → 2n  (∞ + ∞ = ∞)', fontsize=13, fontweight='bold', color='#7c3aed')

for i in range(8):
    is_odd = (i+1) % 2 == 1
    c = '#10b981' if is_odd else '#3b82f6'
    lbl = 'BOŞ' if is_odd else f'{(i+1)//2}'
    rect = patches.FancyBboxPatch((i*1.3, 0.6), 1.1, 0.7, boxstyle="round,pad=0.1",
                                   facecolor=c, alpha=0.8 if is_odd else 0.5, edgecolor='white', lw=1.5)
    ax2.add_patch(rect)
    ax2.text(i*1.3+0.55, 0.95, lbl, ha='center', va='center', fontsize=10, color='white', fontweight='bold')
    ax2.text(i*1.3+0.55, 0.35, f'Oda {i+1}', ha='center', fontsize=8, color='#94a3b8')

plt.tight_layout()
plt.show()
```

### Hilbert Uzayı
Kuantum durumlarının yaşadığı matematiksel ev. Üç özellik:

1. **Vektör uzayı** → süperpozisyon mümkün
2. **İç çarpım** → olasılık hesaplanır (Born kuralı)
3. **Tamlık** → yakınsak dizilerin limiti uzayda kalır (delik yok)

**Boyut = 2ⁿ** → n qubit = 2ⁿ boyutlu Hilbert uzayı

```python
import matplotlib.pyplot as plt
import numpy as np

qubits = [1, 2, 3, 5, 10, 20, 30, 50]
dims = [2**n for n in qubits]
colors = ['#3b82f6','#10b981','#f59e0b','#ef4444','#7c3aed','#0891b2','#d946ef','#dc2626']

fig, ax = plt.subplots(figsize=(10, 5))
bars = ax.bar(range(len(qubits)), dims, color=colors, alpha=0.85, edgecolor='white', lw=1.5)
ax.set_yscale('log')
ax.set_xticks(range(len(qubits)))
ax.set_xticklabels([f'{n} qubit' for n in qubits], fontsize=11)
ax.set_ylabel('Hilbert Uzayı Boyutu (log)', fontsize=12)
ax.set_title('Qubit → Boyut (Üstel Büyüme)', fontsize=14, fontweight='bold')

for bar, n, d in zip(bars, qubits, dims):
    label = f'2^{n}={d}' if d < 1e6 else f'2^{n}≈10^{int(np.log10(d))}'
    ax.text(bar.get_x()+bar.get_width()/2, d*2, label, ha='center', fontsize=9, fontweight='bold')

ax.axhline(y=2**50, color='red', ls='--', alpha=0.4)
ax.text(6.5, 2**51, '← Klasik simülasyon imkansız!', fontsize=10, color='red', fontweight='bold')
ax.grid(axis='y', alpha=0.2)
plt.tight_layout()
plt.show()
```

## 7. Fiziksel ve Mantıksal Qubit'ler

**Fiziksel qubit** = donanımdaki gerçek qubit (gürültülü, hata yapar)

**Mantıksal qubit** = birçok fiziksel qubit'le korunan soyut qubit (~1000 fiziksel = 1 mantıksal)

> Analoji: Fiziksel = kurşun kalemle yazılmış harf (silinebilir). Mantıksal = 100 kopyası (birkaçı silinse bile okunur).

```python
import matplotlib.pyplot as plt

sirket   = ['IBM Eagle', 'IBM Osprey', 'IBM Condor', 'Google\nSycamore', 'Atom\nComputing']
fiziksel = [127, 433, 1121, 53, 1180]
mantiksal = [f/1000 for f in fiziksel]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4.5))

bars1 = ax1.bar(sirket, fiziksel, color='#3b82f6', alpha=0.8, edgecolor='white')
ax1.set_title('Fiziksel Qubit Sayıları', fontsize=13, fontweight='bold')
for b, v in zip(bars1, fiziksel):
    ax1.text(b.get_x()+b.get_width()/2, v+20, str(v), ha='center', fontsize=10, fontweight='bold')

bars2 = ax2.bar(sirket, mantiksal, color='#7c3aed', alpha=0.8, edgecolor='white')
ax2.set_title('Tahmini Mantıksal Qubit (÷1000)', fontsize=13, fontweight='bold')
ax2.axhline(y=1, color='red', ls='--', alpha=0.6)
ax2.text(3.5, 1.05, '← 1 mantıksal qubit bile yok!', fontsize=10, color='red')
for b, v in zip(bars2, mantiksal):
    ax2.text(b.get_x()+b.get_width()/2, v+0.02, f'{v:.2f}', ha='center', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.show()
print("🔑 Bugünkü kuantum bilgisayarlar NISQ çağında → QML bu çağ için tasarlanmış!")
```
